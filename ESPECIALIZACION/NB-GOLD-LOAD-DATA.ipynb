```python
from pyspark.sql import functions as F
```

```python
df = spark.read.table('especializacion.silver_zone.data_kagle')
display(df)
```

```python
df = df.select(
    F.col("id_unico"),
    F.col("id_producto"),
    F.col("nombre_producto"),
    F.col("tipo_producto"),
    F.col("medidas_producto"),
    F.col("descripcion_producto"),
    F.col("categoria_principal"),
    F.col("sub_categoria"),
    F.col("calificacion_producto"),
    F.col("cantidad_calificaciones"),
    F.col("vendible_linea"),
    F.col("url"),
    F.col("precio"),
    F.col("currency"),
    F.col("pais")
)
display(df)
```

```python
df.write.format('delta').mode('overwrite').option('overwriteSchema', 'true').saveAsTable('especializacion.gold_zone.data_kagle')
```
