```python
from pyspark.sql import functions as F
```

```python
df = spark.read.table('especializacion.bronze_zone.data_kagle')
display(df)
```

```python
dict_headers_es = {
    'unique_id': 'id_unico',
    'product_id': 'id_producto',
    'product_name': 'nombre_producto',
    'product_type': 'tipo_producto',
    'product_measurements': 'medidas_producto',
    'product_description': 'descripcion_producto',
    'main_category': 'categoria_principal',
    'sub_category': 'sub_categoria',
    'product_rating': 'calificacion_producto',
    'product_rating_count': 'cantidad_calificaciones',
    'badge': 'badge',
    'online_sellable': 'vendible_linea',
    'price': 'precio',
    'discount': 'descuento',
    'sale_tag': 'etiqueta_venta',
    'country': 'pais'
}

for old_name, new_name in dict_headers_es.items():
    df = df.withColumnRenamed(old_name, new_name)
display(df)
```

```python
df = df.replace(["None", "NA", "NONE",'none'], None)
display(df)
```

```python
df = df \
    .withColumn('calificacion_producto', F.col('calificacion_producto').cast('double')) \
    .withColumn('cantidad_calificaciones', F.col('cantidad_calificaciones').cast('double')) \
    .withColumn('id_unico', F.col('id_unico').cast('string')) \
    .withColumn('id_producto', F.col('id_producto').cast('string'))
display(df)
```

```python
df.write.format('delta').mode('overwrite').option('overwriteSchema', 'true').saveAsTable('especializacion.silver_zone.data_kagle')
```
